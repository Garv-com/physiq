```python
import sys
sys.path.append("..")

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/synthetic_workout_log.csv")
df.head(15)
```

```python
squat_df = df[df["exercise"] == "Squat"]

plt.figure(figsize=(10, 5))
plt.scatter(squat_df["date"], squat_df["weight_kg"], alpha=0.4, label="Observed set weights")
plt.plot(squat_df["date"], squat_df["true_top_weight"], color="green", linewidth=2, linestyle="--", label="True top weight")
plt.xlabel("Date")
plt.ylabel("Weight (kg)")
plt.title("Squat Progression: Observed Sets vs True Top Weight")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
